```python
import numpy as np

for d in [2, 10, 100, 1000, 10000]:
    dists = []
    for _ in range(10000):
        u = np.random.randn(d); u /= np.linalg.norm(u)
        v = np.random.randn(d); v /= np.linalg.norm(v)
        dists.append(np.linalg.norm(u - v))
    dists = np.array(dists)
    print(f"d={d:6d} | mean={dists.mean():.4f} | std={dists.std():.4f} "
          f"| min={dists.min():.4f} | max={dists.max():.4f}")
```
